# Phase 8 EDA
## 1. Phase 8 Overview
This project uses synthetic/anonymized data for portfolio and analytical demonstration purposes.
Phase 8 performs descriptive and diagnostic analysis only; no predictive model is trained in this phase.

## 2. Environment / Imports

```
import pandas as pd
import importlib
import sys
import os
import json
sys.path.insert(0, os.path.abspath('python/eda'))
```

## 3. Data Extraction

```
m01 = importlib.import_module('01_data_extraction')
df_calls = m01.extract_call_level()
df_conv = m01.extract_conversation_level()
df_cust = m01.create_customer_dataset(df_calls, df_conv)
print(f'Calls: {len(df_calls)}, Conversations: {len(df_conv)}, Customers: {len(df_cust)}')
```

## 4. Data Quality

```
m02 = importlib.import_module('02_data_quality_profile')
profile = m02.run_quality_profile(df_calls, df_conv, df_cust)
print(f"Validation complete. Nulls in connection status: {profile['fact_calls_quality']['nulls']['connection_status']}")
```

## 5. Univariate Analysis

```
m03 = importlib.import_module('03_univariate_analysis')
univariate = m03.run_univariate(df_calls, df_conv)
print(list(univariate['numerical_summary'].keys()))
```

## 6. Bivariate Analysis

```
m04 = importlib.import_module('04_bivariate_analysis')
bivariate = m04.run_bivariate(df_calls, df_conv)
print("Bivariate generated successfully")
```

## 7. Conversational AI

```
m05 = importlib.import_module('05_conversational_ai_analysis')
ai_stats = m05.run_conversational_ai(df_calls, df_conv)
print(f"Intent accuracy: {ai_stats.get('intent_accuracy_overall', 'N/A')}")
```

## 8. Customer Journey

```
tasks_started = df_calls['task_started_flag'].sum()
tasks_completed = df_calls['task_completed_flag'].sum()
print(f"Tasks started: {tasks_started}, Tasks completed: {tasks_completed}")
```

## 9. Collections

```
m06 = importlib.import_module('06_collections_analysis')
col_stats = m06.run_collections(df_calls)
print(f"PTP conversion rate: {col_stats.get('ptp_conversion_rate', 'N/A')}")
```

## 10. Customer Behavior

```
m07 = importlib.import_module('07_customer_behavior_analysis')
cust_stats = m07.run_customer_behavior(df_cust)
print(f"Total connected calls: {cust_stats['metrics_summary']['connected_calls']}")
```

## 11. Temporal Analysis

```
m08 = importlib.import_module('08_temporal_analysis')
temp_stats = m08.run_temporal(df_calls)
print(f"Temporal output dict keys: {list(temp_stats.keys())}")
```

## 12. Escalation Association

```
m09 = importlib.import_module('09_escalation_analysis')
esc_stats = m09.run_escalation(df_calls, df_conv)
print(f"Escalation mapped for {esc_stats['merged_unique']} unique calls")
```

## 13. Correlation

```
m10 = importlib.import_module('10_correlation_analysis')
corr_stats = m10.run_correlation(df_calls, df_conv)
for k, v in corr_stats.items():
    print(f"{k}: {v['interpretation']} (r={v['coefficient']:.3f})")
```

## 14. Outliers

```
m11 = importlib.import_module('11_outlier_analysis')
outliers = m11.run_outliers(df_calls)
print("Outlier analysis generated.")
```

## 15. Statistical Tests

```
# Print tests from JSON report
with open('reports/phase8_eda_report.json', 'r') as f:
    r_json = json.load(f)
for t in r_json['statistical_tests']:
    if t['status'] == 'VALID':
        print(f"{t['test_name']} | {t['variables']} | n={t['n']} | stat={t['statistic']:.4f} | p={t['p_value']:.4f} | {t['interpretation']}")
```

## 16. ML Readiness

```
m12 = importlib.import_module('12_ml_readiness')
ml_inv = m12.run_ml_readiness()
print(f"Allowed features: {len(ml_inv['allowed'])}, Prohibited features: {len(ml_inv['prohibited'])}")
```

## 17. Key Findings

```
for fnd in r_json['findings']:
    print(f"- {fnd['category']}: {fnd['Observation']}")
```
